# Loan Data Cleaning

Data cleaning pipeline for `loan_data_2007_2014.csv` (Lending Club loans, 2007–2014).

```python
import pandas as pd
import numpy as np

DATA_PATH = "loan_data_2007_2014.csv"

df = pd.read_csv(DATA_PATH, low_memory=False)

print(f"Rows: {df.shape[0]:,}  |  Columns: {df.shape[1]}")
print(f"Memory usage: {df.memory_usage(deep=True).sum() / 1024**2:.1f} MB")
df.head()
```

## 1. Initial inspection

```python
missing_summary = (
    pd.DataFrame({
        "missing_count": df.isna().sum(),
        "missing_pct": (df.isna().mean() * 100).round(2),
        "dtype": df.dtypes.astype(str),
    })
    .sort_values("missing_pct", ascending=False)
)

print("Columns with any missing values:")
display(missing_summary[missing_summary["missing_count"] > 0])
```

## 2. Remove useless columns

Drop columns that contain no usable data:
- CSV export index column (`Unnamed: 0`)
- Columns that are 100% missing (no values at all)

```python
# Drop CSV index artifact
if "Unnamed: 0" in df.columns:
    df = df.drop(columns=["Unnamed: 0"])

# Identify columns with zero non-null values
empty_columns = df.columns[df.isna().all()].tolist()
print(f"Removing {len(empty_columns)} empty columns (100% missing):")
print(empty_columns)

df = df.drop(columns=empty_columns)

print(f"\nShape after dropping empty columns: {df.shape}")
```

```python
df.info()
```

## 3. Standardize missing values

Treat blank strings and whitespace-only values as missing so counts and imputation are consistent.

```python
text_columns = df.select_dtypes(include=["object", "string"]).columns

for col in text_columns:
    df[col] = df[col].replace(r"^\s*$", np.nan, regex=True)

print(f"Standardized blanks to NaN in {len(text_columns)} text columns.")
```

## 4. Fix data types

Convert date-like and numeric columns to appropriate types where possible.

```python
# Term is stored as " 36 months" / " 60 months" — keep the numeric part
df["term"] = df["term"].astype(str).str.extract(r"(\d+)").astype(float)

# Parse date columns (month-year format from Lending Club)
date_columns = [
    "issue_d",
    "earliest_cr_line",
    "last_pymnt_d",
    "next_pymnt_d",
    "last_credit_pull_d",
]

for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], format="%b-%y", errors="coerce")

# Ensure id columns are integers
for col in ["id", "member_id"]:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce").astype("Int64")

print("Updated dtypes:")
print(df.dtypes.value_counts())
```

## 6. Final summary

```python
final_missing = (
    pd.DataFrame({
        "missing_count": df.isna().sum(),
        "missing_pct": (df.isna().mean() * 100).round(2),
    })
    .query("missing_count > 0")
    .sort_values("missing_pct", ascending=False)
)

print("Remaining columns with missing values:")
display(final_missing)

print("\nCleaned dataset preview:")
display(df.head())

# `df` is ready for analysis / modeling
```

```python
df.info()
```

## 7. Binary target label (`label`)

For default prediction we only keep loans with a **final** outcome. Ongoing or in-progress statuses are dropped because the label is not yet known.

| `loan_status` | Action |
|---|---|
| Current | **Remove** — loan still active |
| In Grace Period | **Remove** — still in early repayment |
| Late (16–30 days), Late (31–120 days) | **Remove** — delinquent but outcome not final |
| Fully Paid | **label = 1** (good) |
| Charged Off, Default | **label = 0** (bad) |
| Does not meet credit policy (Fully Paid / Charged Off) | **Keep** — rare policy exceptions with a resolved outcome |

```python
# Statuses with unknown / in-progress outcome — drop before labeling
DROP_STATUSES = [
    "Current",
    "In Grace Period",
    "Late (16-30 days)",
]

rows_before = len(df)
df = df[~df["loan_status"].isin(DROP_STATUSES)].copy()
print(f"Dropped {rows_before - len(df):,} in-progress rows")
print(f"Remaining rows: {len(df):,}")
print(df["loan_status"].value_counts())
```

```python
# 1 = good debt (repaid), 0 = bad debt (default / charged off)
GOOD_STATUSES = [
    "Fully Paid",
    "Does not meet the credit policy. Status:Fully Paid",
]
BAD_STATUSES = [
    "Charged Off",
    "Default",
    "Does not meet the credit policy. Status:Charged Off",
    "Late (31-120 days)"
]

status_to_label = {s: 1 for s in GOOD_STATUSES} | {s: 0 for s in BAD_STATUSES}
df["label"] = df["loan_status"].map(status_to_label)

unmapped = df["label"].isna().sum()
if unmapped:
    print("Unmapped statuses (should be 0):")
    print(df.loc[df["label"].isna(), "loan_status"].value_counts())
else:
    print("All remaining rows mapped to label.")

print("\nLabel distribution:")
print(df["label"].value_counts().rename({0: "bad (0)", 1: "good (1)"}))
print(f"\nBad rate: {(df['label'] == 0).mean():.2%}")
```

## 8. Feature engineering

Transform selected categorical / text columns into model-ready features.

```python
df['title'].value_counts()
```

```python
# desc → has_desc (1 if borrower wrote a description, else 0)
df["has_desc"] = df["desc"].notna().astype(int)
df = df.drop(columns=["desc"])

# emp_length → numeric years
# Max at 10+ years
def parse_emp_length(value):
    if pd.isna(value):
        return np.nan
    text = str(value).strip().lower()
    if text == "< 1 year":
        return 0.5
    if "+" in text:
        return float(text.split("+")[0])
    return float(text.split()[0])

df["emp_length"] = df["emp_length"].map(parse_emp_length)

# home_ownership: merge rare ANY into OTHER
df["home_ownership"] = (
    df["home_ownership"]
    .replace({"ANY": "OTHER"})
    .str.upper()
)

print("has_desc:", df["has_desc"].value_counts().to_dict())
print("emp_length sample:", df["emp_length"].describe()[["min", "50%", "max"]].to_dict())
print("home_ownership:\n", df["home_ownership"].value_counts())
```

```python
df['emp_length'].value_counts()
```

```python
# emp_title → emp_pay_tier (low / medium / high) using job-title keywords
HIGH_PAY_KEYWORDS = [
    "doctor", "physician", "surgeon", "dentist", "lawyer", "attorney",
    "engineer", "software", "developer", "architect", "consultant",
    "director", "executive", "ceo", "cfo", "vice president", " vp",
    "manager", "professor", "pharmacist", "analyst", "pilot",
    "nurse practitioner", "financial adviser", "investment",
]
LOW_PAY_KEYWORDS = [
    "cashier", "clerk", "server", "waiter", "waitress", "barista",
    "janitor", "cleaner", "housekeep", "dishwash", "crew member",
    "retail", "sales associate", "stock clerk", "security guard",
    "landscap", "laborer", "warehouse", "picker", "packer",
    "helper", "intern", "student", "delivery driver", "fast food",
    "mcdonald", "burger king", "pizza hut", "subway", "starbucks",
]


def job_pay_tier(title):
    if pd.isna(title):
        return "none"
    text = str(title).lower()
    if any(keyword in text for keyword in HIGH_PAY_KEYWORDS):
        return "high"
    if any(keyword in text for keyword in LOW_PAY_KEYWORDS):
        return "low"
    return "medium"


df["emp_pay_tier"] = df["emp_title"].apply(job_pay_tier)
df = df.drop(columns=["emp_title"])

print(df["emp_pay_tier"].value_counts(dropna=False))
```

## 9. Drop unnecessary, leaky, and sparse columns

### Why some columns have little data

| Column | What it means | Why so much missing |
|--------|---------------|---------------------|
| `tot_coll_amt` | Total amount owed on accounts in collections | **Not collected until ~2013** — 100% missing for 2007–2011 loans |
| `total_rev_hi_lim` | Total revolving credit limit across accounts | Same — new field added mid-dataset |
| `tot_cur_bal` | Total current balance on all credit accounts | Same — new field added mid-dataset |
| `mths_since_last_delinq` | Months since last delinquency on credit report | **Missing = no delinquency on record** (not a data error) |
| `mths_since_last_record` | Months since last public record (bankruptcy, lien, etc.) | **Missing = no public record** |
| `mths_since_last_major_derog` | Months since last major derogatory mark | **Missing = never had one** (~79% missing) |
| `next_pymnt_d` | Next scheduled payment date | **Missing when loan is finished** — paid off or charged off loans have no upcoming payment |

### Columns removed in this step

**Leakage / not usable at loan time:** IDs, URLs, raw target (`loan_status`), payment/outcome fields recorded after origination.

**Dropped (sparse / year-gaps only):** `tot_coll_amt`, `total_rev_hi_lim`, `tot_cur_bal`.

**Kept & engineered (see cell below):** `mths_since_last_delinq`, `mths_since_last_record`, `mths_since_last_major_derog`, `next_pymnt_d` — NaN is recoded to `"None"` meaning the event did not occur (or no upcoming payment), not missing data.

```python
# NaN here means "no event" — encode explicitly as "None" (not missing data)
NONE_LABEL = "None"

MTHS_EVENT_COLS = [
    "mths_since_last_delinq",
    "mths_since_last_record",
    "mths_since_last_major_derog",
]


def format_months_since(value):
    if pd.isna(value):
        return NONE_LABEL
    num = float(value)
    return str(int(num)) if num.is_integer() else str(num)


for col in MTHS_EVENT_COLS:
    df[col] = df[col].apply(format_months_since)

# Next payment date: keep ISO date string when scheduled, else "None"
df["next_pymnt_d"] = df["next_pymnt_d"].apply(
    lambda x: pd.Timestamp(x).strftime("%Y-%m-%d") if pd.notna(x) else NONE_LABEL
)

print("Value counts after encoding (None = event did not apply):")
for col in MTHS_EVENT_COLS + ["next_pymnt_d"]:
    print(f"\n{col}:")
    print(df[col].value_counts().head(5))
    none_count = (df[col] == NONE_LABEL).sum()
    print(f"  '{NONE_LABEL}': {none_count:,} ({none_count / len(df):.1%})")
```

```python
LEAKAGE_OR_UNNECESSARY = [
    # IDs & links
    "id",
    "member_id",
    "url",
    # Raw text / duplicate of purpose (desc, emp_title already engineered)
    "title",
    # Target already encoded in label
    "loan_status",
    # Near-duplicates of loan_amnt
    "funded_amnt",
    "funded_amnt_inv",
    # Known only after loan performance unfolds
    "out_prncp",
    "out_prncp_inv",
    "total_pymnt",
    "total_pymnt_inv",
    "total_rec_prncp",
    "total_rec_int",
    "total_rec_late_fee",
    "recoveries",
    "collection_recovery_fee",
    "last_pymnt_d",
    "last_pymnt_amnt",
    "last_credit_pull_d",
]

# Only balance fields with year-gaps; mths_* / next_pymnt_d engineered in cell above
SPARSE_OR_LOW_VALUE = [
    "tot_coll_amt",
    "total_rev_hi_lim",
    "tot_cur_bal",
]

DROP_COLUMNS = LEAKAGE_OR_UNNECESSARY + SPARSE_OR_LOW_VALUE
cols_to_drop = [c for c in DROP_COLUMNS if c in df.columns]

print(f"Dropping {len(cols_to_drop)} columns:")
for col in cols_to_drop:
    missing_pct = df[col].isna().mean() * 100
    print(f"  {col}: {missing_pct:.1f}% missing")

df = df.drop(columns=cols_to_drop)
print(f"\nShape after drops: {df.shape}")
print(f"Columns remaining: {df.shape[1]}")
```

## 10. Pre-model type cleanup

Prepare columns for modeling now; full encoding/standardization comes later.

```python
NONE_LABEL = "None"  # from event-encoding cell above

# --- Drop: no variance or redundant ---
df = df.drop(columns=["pymnt_plan", "application_type"], errors="ignore")

# --- Datetime -> integer (days since 1970-01-01; nullable Int64) ---
EPOCH = pd.Timestamp("1970-01-01")

for col in ["issue_d", "earliest_cr_line"]:
    if col in df.columns:
        df[col] = ((df[col] - EPOCH).dt.days).astype("Int64")

# --- Months-since columns: "None" -> -1, else integer months ---
MTHS_EVENT_COLS = [
    "mths_since_last_delinq",
    "mths_since_last_record",
    "mths_since_last_major_derog",
]

for col in MTHS_EVENT_COLS:
    df[col] = df[col].apply(
        lambda x: -1 if x == NONE_LABEL else int(float(x))
    ).astype("int64")

# --- initial_list_status: w (whole) = 1, f (fractional) = 0 ---
df["initial_list_status"] = (df["initial_list_status"].str.lower() == "w").astype("int64")

# --- next_pymnt_d: "None" = -1 (no upcoming payment), else days since epoch ---
df["next_pymnt_d"] = df["next_pymnt_d"].apply(
    lambda x: -1
    if x == NONE_LABEL
    else int((pd.Timestamp(x) - EPOCH).days)
).astype("int64")

# Drop zip code
df = df.drop(columns=["zip_code"], errors="ignore")

print("Dropped: pymnt_plan, application_type")
print("\nDtypes after cleanup:")
print(df.dtypes.value_counts())
print("\nSample engineered values:")
print(df[MTHS_EVENT_COLS + ["initial_list_status", "next_pymnt_d", "issue_d"]].head())
```
